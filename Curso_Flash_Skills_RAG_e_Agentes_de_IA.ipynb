```python
import os
from google.colab import userdata

# Carrega a chave da Groq
os.environ['GROQ_API_KEY'] = userdata.get('GROQ_API_KEY')

# Carrega a chave do Hugging Face (opcional, mas recomendado)
os.environ['HF_TOKEN'] = userdata.get('HF_TOKEN')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

# Baixa e prepara o modelo de vetores na memória da GPU
embed_model = HuggingFaceEmbeddings(model_name="mixedbread-ai/mxbai-embed-large-v1")
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.vectorstores import InMemoryVectorStore

def carrega_pdf(caminho_arquivo: str):
    # 1. Instancia o leitor para o PDF informado
    loader = PyPDFLoader(caminho_arquivo)

    # 2. Carrega todas as páginas
    pages = []
    for page in loader.load():
        pages.append(page)

    # 3. Transforma o texto das páginas em vetores e salva no Vector Store
    vectorstore = InMemoryVectorStore.from_documents(pages, embed_model)

    print(f"Arquivo '{caminho_arquivo}' processado com {len(pages)} páginas.")
    return vectorstore
```

```python
# Cria uma base vetorial exclusiva para as propostas do Lula
vector_store_lula = carrega_pdf("plano_lula.pdf")

# Cria uma base vetorial exclusiva para as propostas do Flávio Bolsonaro
vector_store_flavio = carrega_pdf("plano_flavio.pdf")
```

```python
# Busca os 2 trechos mais relevantes sobre economia no plano do Lula
resultados_lula = vector_store_lula.similarity_search("propostas para a economia e inflação", k=2)

for doc in resultados_lula:
    print(f"Página {doc.metadata.get('page', '?')}:")
    print(doc.page_content[:300]) # Exibe os primeiros 300 caracteres
    print("-" * 50)
```

```python
from langchain_core.tools import tool

@tool
def consulta_plano_lula(query: str) -> str:
    """Consulta o plano de governo de Lula para obter propostas e diretrizes sobre um tema específico."""
    retriever = vector_store_lula.as_retriever()
    resultado = retriever.invoke(query)
    return str(resultado)

@tool
def consulta_plano_flavio(query: str) -> str:
    """Consulta o plano de governo de Flávio Bolsonaro para obter propostas e diretrizes sobre um tema específico."""
    retriever = vector_store_flavio.as_retriever()
    resultado = retriever.invoke(query)
    return str(resultado)

# Agrupamos as ferramentas em uma lista
tools = [consulta_plano_lula, consulta_plano_flavio]
```

```python
teste_retorno = consulta_plano_flavio.invoke("segurança pública e combate ao crime")
print(teste_retorno[:400])
```

```python
from langgraph.prebuilt import create_react_agent, ToolNode, tools_condition
from langgraph.checkpoint.memory import MemorySaver
from langgraph.graph import StateGraph, MessagesState, START, END
from langchain_core.messages import HumanMessage, SystemMessage
from IPython.display import Image, display
```

```python
from langchain_groq import ChatGroq

# Instância com o modelo configurado anteriormente
llm = ChatGroq(
    model='openai/gpt-oss-120b',
    temperature=0.2
)

# Instruções claras para o Agente em português
system_prompt = """Você é um assistente imparcial e especializado em análise de propostas políticas e planos de governo das eleições de 2026.
Sua função é responder às dúvidas com base estritamente no contexto obtido pelas ferramentas. Se a informação não constar nos documentos, diga claramente que não encontrou nos planos.

Ferramentas disponíveis:
- consulta_plano_lula: Use para consultar o plano de governo de Lula sobre qualquer área temática.
- consulta_plano_flavio: Use para consultar o plano de governo de Flávio Bolsonaro sobre qualquer área temática.

Se a pergunta do usuário solicitar uma comparação entre os dois candidatos, você deve consultar ambas as ferramentas antes de elaborar a resposta.
"""
```

```python
# Criação do agente com ReAct
agente_planos = create_react_agent(
    model=llm,
    tools=tools,
    prompt=system_prompt
)
```

```python
# 1. Instancia o grafo baseado no estado de mensagens
grafo = StateGraph(MessagesState)

# 2. Adiciona os nós
grafo.add_node("assistente", agente_planos)
grafo.add_node("tools", ToolNode(tools))

# 3. Define as conexões (Arestas)
grafo.add_edge(START, "assistente")
grafo.add_conditional_edges("assistente", tools_condition)
grafo.add_edge("tools", "assistente")
grafo.add_edge("assistente", END)

# 4. Configura a persistência de memória
memoria = MemorySaver()

# 5. Compila o grafo gerando a aplicação final
app = grafo.compile(checkpointer=memoria)
```

```python
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
def chat_com_memoria(mensagem_usuario: str, thread_id="sessao_1", verbose=False):
    config = {"configurable": {"thread_id": thread_id}}

    # Invoca o grafo enviando a mensagem do usuário
    resposta_grafo = app.invoke(
        {"messages": [HumanMessage(content=mensagem_usuario)]},
        config=config
    )

    if verbose:
        # Exibe todo o raciocínio e chamadas de ferramentas
        for msg in resposta_grafo["messages"]:
            msg.pretty_print()
    else:
        # Exibe diretamente a resposta sintetizada final da IA
        resposta_grafo["messages"][-1].pretty_print()
```

```python
chat_com_memoria(
    mensagem_usuario="Quais são as principais propostas do Lula para a área de educação?",
    thread_id="conversa_eleicoes",
    verbose=False
)
```

```python
chat_com_memoria(
    mensagem_usuario="E em relação ao plano do Flávio Bolsonaro sobre esse mesmo assunto?",
    thread_id="conversa_eleicoes",
    verbose=False
)
```
